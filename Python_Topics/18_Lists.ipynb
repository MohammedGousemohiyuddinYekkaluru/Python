## List

```python
# Differences between List and Array
# Array is homogenous means only stores one data type
# List is hetrogenous means can store more than one data type
# Arrays are much faster
# Lists are more programmer friendly
```

```python
# Create a List
l = [] # Empty List
l
```

```python
l = [1,2, 3, 4, 5] # Homogenous List
l
```

```python
l2 = ['hello', 3, 4.5, True, 5+6j] # Hetrogenous List
l2
```

```python
L3 = [1, 2, 3, [4, 5]] # Multi-dimensional List(2D)
L3
```

```python
L4 = [[[1,2], [3,4]], [[5,6], [7,8]]] # 3D
L4
```

```python
L5 = list("Andhra")
L5
```

```python
L6 = list()
L6
```

#### Adding new Elements

```python
l
```

```python
l.append(600) # adds new element at the end of the list
l
```

```python
l.append("hello")
l
```

```python
l.extend([700, 800, 900])  # adds multiple elements at the end of the list
l
```

```python
l.append([5,6])  # append will always try to append one item
l
```

```python
l.extend("goa")  # Extend will always try to append multiple items
l
```

```python
l.insert(1, "world") # Adding items at desired index
l
```
