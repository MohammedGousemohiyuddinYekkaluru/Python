## Tuples

#### Tuples are just like a List with some differences

```python
# Create
T1 = () # Empty Tuple
T1
```

```python
T2 = (1, 2, 3, 4, 5) # Homogenous Tuple
T2
```

```python
T3 = ("Hello", 4, 5.6) # Hetrogenous Tuple
T3
```

```python
T4 = (1, 2, 3, (4, 5))
T4
```

```python
T5 = (1) #single element tuple is not possible in this way
T5
```

```python
type(T5)
```

```python
t5 = ("hello")
t5
```

```python
type(t5)
```

```python
T5 = ("Hello",) # This is how we can create a single element tuple
T5
```

```python
T6 = tuple("goa")
T6
```

```python
T6 = tuple([1, 2, 3, 4, 5])
T6
```
