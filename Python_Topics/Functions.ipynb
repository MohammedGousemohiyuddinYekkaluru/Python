##### Let's create a function(with docstring)

```python
def is_even(num):
    """
  This function returns if a given number is odd or even
  input - any valid integer
  output - odd/even
  created on - 27th may 2026
  """
    if type(num) == int:
        if num % 2 == 0:
            return "even"
        
        else:
            return "odd"
    
    else:
        return "enter valid integer"
```

```python
for i in range(1, 11):
    x = is_even(i)
    print(x)
```

```python
is_even('Hello')
```

#### Types of Arguments

- Default Argument
- Positional Argument
- Keyword Argument

```python
def power(a=1, b=1):
    return a**b
```

```python
# default argument
power()  # using default arguments
```

```python
# positional argument
power(2, 3)
```

```python
# keyword argument
power(b=3, a=2)
```

##### `*args and **kwargs`
`*args and **kwargs` are special Python keywords that are used to pass the variable length of arguments to a function

```python
# *args
# allows us to pass a variable number of non-keyword arguments to a function.

def multiply(*args):
    product = 1

    for i in args:
        product *= i
    
    print(args)  #types of args are tuples
    return product
```

```python
multiply(1, 2, 3, 4, 5, 6, 7)
```

```python
# **kwargs
# **kwargs allows us to pass any number of keyword arguments.
# Keyword arguments mean that they contain a key-value pair, like a Python dictionary.

def display(**kwargs):
    
    for (key, value) in kwargs.items():
        print(key, '->', value)
```

```python
display(india='delhi',srilanka='colombo',nepal='kathmandu')
```

##### Points to remember while using `*args and **kwargs`

- order of the arguments matter(normal -> `*args` -> `**kwargs`)
- The words “args” and “kwargs” are only a convention, you can use any name of your choice
