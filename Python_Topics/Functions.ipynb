##### Let's create a function(with docstring)

```python
def is_even(num):
    """
  This function returns if a given number is odd or even
  input - any valid integer
  output - odd/even
  created on - 27th may 2026
  """
    if type(num) == int:
        if num % 2 == 0:
            return "even"
        
        else:
            return "odd"
    
    else:
        return "enter valid integer"
```

```python
for i in range(1, 11):
    x = is_even(i)
    print(x)
```

##### `*args and **kwargs`
`*args and **kwargs` are special Python keywords that are used to pass the variable length of arguments to a function

```python
# *args
# allows us to pass a variable number of non-keyword arguments to a function.

def multiply(*args):
    product = 1

    for i in args:
        product *= i
    
    print(args)  #types of args are tuples
    return product
```

```python
multiply(1, 2, 3, 4, 5, 6, 7)
```

```python
# **kwargs
# **kwargs allows us to pass any number of keyword arguments.
# Keyword arguments mean that they contain a key-value pair, like a Python dictionary.

def display(**kwargs):
    
    for (key, value) in kwargs.items():
        print(key, '->', value)
```

```python
display(india='delhi',srilanka='colombo',nepal='kathmandu')
```

##### Points to remember while using `*args and **kwargs`

- order of the arguments matter(normal -> `*args` -> `**kwargs`)
- The words “args” and “kwargs” are only a convention, you can use any name of your choice

##### variable scope
- global variable
- local variable

```python
def g(y):
    print(x)
    print(x+1)
x = 5
g(x)
print(x)
```

```python
def f(y):
    x = 1
    x += 1
    print(x)
x = 5
f(x)
print(x)
```

```python
def h(y):
    x += 1    # editing the global variable from the function is not allowed and not a good practice
x = 5
h(x)
print(x)
```

```python
def f(x):
   x = x + 1
   print('in f(x): x =', x)
   return x

x = 3
z = f(x)
print('in main program scope: z =', z)
print('in main program scope: x =', x)
```

##### Nested Functions

```python
def f():
    def g():
        print("inside function g")
    g()
    print("inside functon f")
```

```python
f()
```

```python
def g(x):
    def h():
        x = 'abc'
    x = x + 1
    print('in g(x): x =', x)
    h()
    return x

x = 3
z = g(x)
```

```python
def g(x):
    def h(x):
        x = x+1
        print("in h(x): x = ", x)
    x = x + 1
    print('in g(x): x = ', x)
    h(x)
    return x

x = 3
z = g(x)
print('in main program scope: x = ', x)
print('in main program scope: z = ', z)
```
