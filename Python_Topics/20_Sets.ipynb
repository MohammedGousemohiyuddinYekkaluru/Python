## SETS

- 1) Sets do not allow duplicates
- 2) Sets have no indexing/Slicing
- 3) Sets dont allow mutable data types
- 4) Set itself is a mutable data type
- 5) that's why 2D, 3D sets are not possible

```python
S1 = {}  # You cant create empty set like this, this is for dictionary
S1
```

```python
type(S1)
```

```python
S1 = set()
S1
```

```python
type(S1)
```

```python
S2 = {1, 2, 3, 4, 5}  # Homogenous Set
S2
```

```python
S3 = {"Hello", 4.5, True}  # Hetrogenous Set
S3
```

```python
S4 = {1, 1, 2, 2, 3, 3}  # no duplicates are allowed
S4
```

### Set Operations

```python
S1 = {1, 2, 3, 4, 5}
S2 = {3, 4, 5, 6, 7}
```

```python
S1 + S2    # You cannot concatinate sets
```

```python
S1 * 3   # You cannot multiply them
```

```python
for i in S1:   # Loops are possible
    print(i)
```

```python
1 in S1   # Membership is also possible
```
