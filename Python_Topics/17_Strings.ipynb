#### Accessing Substrings from a String

```python
# Concept of Indexing
c = "hello"
print(c)
print(c[0])
print(c[1])
print(c[2])
print(c[3])
print(c[4])
```

```python
# Types of Indexing
# Positive Indexing
# Negative Indexing

print(c[-1]) # helpfu to access last element of a string
print(c[-2])
```

```python
# Slicing
c = "Hello World"
print(c)
```

```python
print(c[0:4]) #slicing [start:end(not included)]
print(c[0:5])
```

```python
print(c[:])
```

```python
print(c[:4])
print(c[2:6:2])
print(c[2:])
```

```python
print(c[0:8:3])
```

```python
c = "gouse world"
print(c[0:8:3])
```

```python
print(c[::-1]) # reverse
```

```python
print(c[-1:-6:-1])
```
