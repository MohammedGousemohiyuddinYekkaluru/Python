## Dictionary

```python
{"Name": "Gouse", "Gender": "Male"}
```

- 1) Dictionary has no indexing
- 2) Dictionary is a mutable types
- 3) keys -> immutable
- 4) Values -> they can be mutable
- 5) Keys should be unique

#### Mutable data types => means in the same memory location you can edit the items
- List
- Sets
- Dictionary

#### Immutable data types
- Strings
- Tuples
- int
- float
- Boolean
- Complex

```python
D = {}
D
```

```python
D = {"Name": "Gouse", "Gender": "Male"}
D
```
