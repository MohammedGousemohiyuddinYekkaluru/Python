#### Q2:- Write a program that will convert celsius value to fahrenheit.

````python
temp_in_celsius = int(input("Enter Temperature in celsius : "))

temp_in_fahrenheit = (temp_in_celsius * 1.8) + 32

print(f"{temp_in_celsius}°C => {temp_in_fahrenheit}°F")
````

#### Q3:- Take 2 numbers as input from the user.Write a program to swap the numbers without using any special python syntax.

````python
first_num = int(input("Enter first number : "))
second_num = int(input("Enter Second number : "))

print(f"first_num -> {first_num}, second_num -> {second_num}")

temp = first_num
first_num = second_num
second_num = temp

print(f"first_num -> {first_num}, second_num -> {second_num}")
````

#### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python
x_1 = int(input("Enter x_1 : "))
x_2 = int(input("Enter x_2 :"))
y_1 = int(input("Enter y_1 :"))
y_2 = int(input("Enter y_2 :"))

eucli_dist = ((x_2 - x_1) ** 2 + (y_2 - y_1) ** 2) ** 0.5

print(f"Euclidean distance between {x_1, y_1} & {x_2, y_2} is {eucli_dist}")
````

#### Q5:- Write a program to find the simple interest when the value of principle,rate of interest and time period is provided by the user.

````python
principle = int(input("Enter the Principle : "))
roi = int(input("Enter rate of interest : "))
time_period = int(input("Enter time period : "))

simple_interest = (principle * time_period * roi) / 100

print(f"Simple Interest for {principle}rupees, {time_period}years, {roi}% is {simple_interest}")
````
