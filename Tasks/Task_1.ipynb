#### Q2:- Write a program that will convert celsius value to fahrenheit.

````python
temp_in_celsius = int(input("Enter Temperature in celsius : "))

temp_in_fahrenheit = (temp_in_celsius * 1.8) + 32

print(f"{temp_in_celsius}°C => {temp_in_fahrenheit}°F")
````

#### Q3:- Take 2 numbers as input from the user.Write a program to swap the numbers without using any special python syntax.

````python
first_num = int(input("Enter first number : "))
second_num = int(input("Enter Second number : "))

print(f"first_num -> {first_num}, second_num -> {second_num}")

temp = first_num
first_num = second_num
second_num = temp

print(f"first_num -> {first_num}, second_num -> {second_num}")
````

#### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python
x_1 = int(input("Enter x_1 : "))
x_2 = int(input("Enter x_2 :"))
y_1 = int(input("Enter y_1 :"))
y_2 = int(input("Enter y_2 :"))

eucli_dist = ((x_2 - x_1) ** 2 + (y_2 - y_1) ** 2) ** 0.5

print(f"Euclidean distance between {x_1, y_1} & {x_2, y_2} is {eucli_dist}")
````

#### Q5:- Write a program to find the simple interest when the value of principle,rate of interest and time period is provided by the user.

````python
principle = int(input("Enter the Principle : "))
roi = int(input("Enter rate of interest : "))
time_period = int(input("Enter time period : "))

simple_interest = (principle * time_period * roi) / 100

print(f"Simple Interest for {principle}rupees, {time_period}years, {roi}% is {simple_interest}")
````

#### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

````python
n = int(input("Enter a number : "))

sum_of_squares = int((n*(n+1)*(2*n+1)) / 6)

print(f"sum of squares of first {n} numbers is {sum_of_squares}")
````

#### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
first_term = int(input("Enter the first term of the series : "))
second_term = int(input("Enter the second term of the series : "))

nth_term = int(input("Enter which term do you want to know : "))

distance = second_term - first_term

nth_value = first_term + ((nth_term - 1) * distance)

print(f"the value of the nth term in the series is {nth_value}")
````

#### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python
print("Give all the readings in same units cm(centimeter), m(meter), etc.,")

tank_length = int(input("Enter the length of the tank : "))
tank_breadth = int(input("Enter the breadth of the tank : "))
tank_height = int(input("Enter the height of the tank : "))

glass_radius = int(input("Enter the radius of the glass : "))
glass_height = int(input("Enter the height of the glass : "))

volume_of_tank = tank_length * tank_breadth * tank_height
volume_of_glass = 3.14 * (glass_radius**2) * glass_height

import math
N_glasses = math.floor(volume_of_tank / volume_of_glass)

print(f"for Tank length {tank_length}, breadth {tank_breadth}, height {tank_height}")
print(f"and glass height {glass_height} and radius {glass_radius}")
print(f"Total about {N_glasses} glasses of milk can be obtained")
````
