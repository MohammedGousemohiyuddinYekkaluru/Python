##### `Problem 1` : Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

> Salary(Lakhs) : Tax(%)

*   Below 5 : 0%
*   5-10 : 10%
*   10-20 : 20%
*   aboove 20 : 30%

````python
CTC_amount = int(input("Enter your CTC : "))

in_hand_salary = CTC_amount - (CTC_amount * 0.1) - (CTC_amount * 0.05) - (CTC_amount * 0.03)

if CTC_amount > 2000000:
    in_hand_salary -= (in_hand_salary * 0.3)

elif CTC_amount > 1000000:
    in_hand_salary -= (in_hand_salary * 0.2)

elif CTC_amount > 500000:
    in_hand_salary -= (in_hand_salary * 0.1)

print(f"After all deductions your in hand salary is {in_hand_salary}")
````

##### `Problem 6` - Find the factorial of a given number.

Write a program to use the loop to find the factorial of a given number.

The factorial (symbol: `!`) means to multiply all whole numbers from the chosen number down to 1.

For example: calculate the factorial of 5

```bash
5! = 5 × 4 × 3 × 2 × 1 = 120
```

Output: 

```bash
120
```

````python
given_number = int(input("Enter a number : "))

factorial = 1

if given_number < 0:
    print("Can't perform factorial for such number")

else:
    for i in range(2, given_number + 1):
        factorial *= i
    
    print(f"factorial for {given_number} is {factorial}")
````

```

`Output:`

```bash
24567

````python
given_number = int(input("Enter a number : "))
temp = given_number

reverse_num = 0

while temp > 0:
    last_digit = temp % 10
    reverse_num = (reverse_num * 10) + last_digit

    temp //= 10

print(reverse_num)
````

##### `Problem 8` : Take a user input as integer N. Find out the sum from 1 to N. If any number if divisible by 5, then skip that number. And if the sum is greater than 300, don't need to calculate the sum further more. Print the final result. And don't use for loop to solve this problem.

**Example 1:**

`Input:`

```bash
30
```

`Output:`

```bash
276
```

````python
user_input = int(input("Enter a number : "))
start_num = 1

result_sum = 0

while start_num <= user_input:
    if start_num % 5 != 0:
        result_sum += start_num

    if result_sum + start_num > 300:
        break

    start_num += 1

print(result_sum)
````

##### `Problem 9` : Write a program that keeps on accepting a number from the user until the user enters Zero. Display the sum and average of all the numbers.

````python
counter = 0
result_sum = 0

while True:
    user_input = int(input("Enter a number : "))

    if user_input == 0:
        break

    counter += 1
    result_sum += user_input

print(f"sum = {result_sum}")
print(f"avg = {result_sum / counter}")
````

##### `Problem 11` : Write a program, which will find all such numbers between 1000 and 3000 (both included) such that each digit of the number is an even number. The numbers obtained should be printed in a space-separated sequence on a single line.

````python
for i in range(1000, 3001):
    flag = True

    curr = i

    while curr > 0:
        last = curr % 10

        if last % 2 != 0:
            flag = False
            break

        curr //= 10
    
    if flag:
        print(i, end=",")
````

##### `Problem 12` : A robot moves in a plane starting from the original point (0,0). The robot can move toward UP, DOWN, LEFT and RIGHT with a given steps. 
The trace of robot movement is shown as the following:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
> The numbers after the direction are steps. 

> `!` means robot stop there.

**Please write a program to compute the distance from current position after a sequence of movement and original point.** 

*If the distance is a float, then just print the nearest integer.*

Example:

`Input`:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
`Output`:
```
2
```

````python
import math
origin = [0, 0]
pos = [0, 0]

while True:
    movement = input("Enter the robot path : ").upper()

    if movement == "!":
        break

    direction = movement.split()[0]
    steps = int(movement.split()[1])

    if direction == "UP":
        pos[1] += steps

    elif direction == "DOWN":
        pos[1] -= steps

    elif direction == "LEFT":
        pos[0] -= steps

    else:
        pos[0] += steps 

distance = round(math.dist(origin, pos))
print(distance)
````

##### `Problem 13` : Write a program to print whether a given number is a prime number or not

````python
import math

given_number = int(input("Enter a number : "))
is_prime = True

if given_number <= 1:
    is_prime = False

elif given_number == 2:
    is_prime = True

elif given_number %2 == 0:
    is_prime = False

else:
    is_prime = True
    divisor = 3
    limit = int(math.sqrt(given_number))

    while divisor <= limit:
        if given_number % divisor == 0:
            is_prime = False
            break

        divisor += 2

if is_prime:
    print(f"given number {given_number} is a prime number")

else:
    print(f"given number {given_number} is not an prime number")
````

##### `Problem 14` : Print all the Armstrong numbers in a given range.
Range will be provided by the user<br>
Armstrong number is a number that is equal to the sum of cubes of its digits. For example 0, 1, 153, 370, 371 and 407 are the Armstrong numbers.

````python
given_number = int(input("Enter a number : "))

if given_number < 0:
    print("Negative numbers cannot be armstrong numbers.")

else:
    no_of_digits = len(str(given_number))
    temp = given_number
    result_sum = 0

    while temp > 0:
        result_sum += (temp % 10) ** no_of_digits
        temp //= 10

    if given_number == result_sum:
        print(f"{given_number} is an armstrong number")
    
    else:
        print(f"{given_number} is not an armstrong number")
````

##### `Problem 15` : Calculate the angle between the hour hand and minute hand.

Note: There can be two angles between hands; we need to print a minimum of two. Also, we need to print the floor of the final result angle. For example, if the final angle is 10.61, we need to print 10.

Input:<br>
H = 9 , M = 0<br>
Output:<br>
90<br>
Explanation:<br>
The minimum angle between hour and minute
hand when the time is 9 is 90 degress.

````python
hour = int(input("Enter hour : "))
minute = int(input("Enter minutes : "))

hour = hour % 12  # because a clock repeats every 12 hours. 13 => 1

hour_angle = (30 * hour) + (0.5 * minute)  # Hour hand moves 30° per hour and 0.5° per minute
minute_angle = 6 * minute  # Minute hand moves 6° per minute

angle = abs(hour_angle - minute_angle)

final_angle = min(angle, 360 - angle)

print(f"for hour = {hour} and minute = {minute}")
print(f"the angle is {final_angle} degrees")
````
