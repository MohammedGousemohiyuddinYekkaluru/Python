### `Q1:` Join Tuples if similar initial element
While working with Python tuples, we can have a problem in which we need to perform concatenation of records from the similarity of initial element. This problem can have applications in data domains such as Data Science.

For eg.
```
Input  : test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)] 
Output : [(5, 6, 7, 8), (6, 10), (7, 13)] 
```

````python
test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]
unique = []

for i in test_list:
    unique.append(i[0])

unique = set(unique)

result = []
for i in unique:
    result.append([i])

    for j in test_list:
        if j[0] == i:
            result[-1].append(j[1])

print(list(map(tuple, result)))
````

### `Q2:` Multiply Adjacent elements (both side) and take sum of right and lest side multiplication result. 


For eg.
```
The original tuple : (1, 5, 7, 8, 10)
Resultant tuple after multiplication : 

(1*5, 1*5+5*7, 7*5 + 7*8, 8*7 + 8*10, 10*8) -> (5, 40, 91, 136, 80)

output-(5, 40, 91, 136, 80)
```

````python
t = (1, 5, 7, 8, 10)

l = []

l.append(t[0] * t[1])

for i in range(1, len(t)-1):
    l.append(((t[i-1]*t[i]) + (t[i] * t[i+1])))

l.append(t[-1] * t[-2])

result = tuple(l)
print(result)
````

### `Q3`: Check is tuples are same or not?
Two tuples would be same if both tuples have same element at same index
```
t1 = (1,2,3,0)
t2 = (0,1,2,3)

t1 and t2 are not same
```

````python
t1 = (1,2,3,0)
t2 = (0,1,2,3)

is_same = True

for i, j in zip(t1, t2):
    if i != j:
        is_same = False
        break

if is_same:
    print("t1 and t2 are same")

else:
    print("t1 and t2 are not same")
````

### `Q4`: Count no of tuples, list and set from a list
```
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]

```
`Output:`

```
List-2
Set-2
Tuples-1
```

````python
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]

n_list = 0
n_set = 0
n_tuple = 0

for i in list1:
    type_of_el = type(i)
    
    if type_of_el == set:
        n_set += 1

    elif type_of_el == list:
        n_list += 1
    
    elif type_of_el == tuple:
        n_tuple += 1

print("List-",n_list)
print("Set-",n_set)
print("Tuples-",n_tuple)
````

````python
# Another way
output = [0, 0, 0]

for i in list1:
    type_of_el = type(i)
    
    if type_of_el == set:
        output[0] += 1

    elif type_of_el == list:
        output[1] += 1
    
    elif type_of_el == tuple:
        output[2] += 1

print('List-{}\nSet-{}\nTuples-{}'.format(output[0], output[1], output[2]))
````

### `Q1:` Write a program to find set of common elements in three lists using sets.
```
Input : ar1 = [1, 5, 10, 20, 40, 80]
        ar2 = [6, 7, 20, 80, 100]
        ar3 = [3, 4, 15, 20, 30, 70, 80, 120]

Output : [80, 20]
```

````python
ar1 = [1, 5, 10, 20, 40, 80]
ar2 = [6, 7, 20, 80, 100]
ar3 = [3, 4, 15, 20, 30, 70, 80, 120]

final_ar = ar1 + ar2 + ar3

common_list = []

for i in final_ar:
    if (i in ar1 and i in ar2 and i in ar3):
        common_list.append(i)

output_list = list(set(common_list))
print(output_list)
````

````python
# Another method
s1 = set(ar1)
s2 = set(ar2)
s3 = set(ar3)

result = list((s1 & s2) & s3)

print(result)
````

### `Q4`: find union of n arrays.

**Example 1:**

Input:
```bash
[[1, 2, 2, 4, 3, 6],
 [5, 1, 3, 4],
 [9, 5, 7, 1],
 [2, 4, 1, 3]]
```

Output:

```bash
[1, 2, 3, 4, 5, 6, 7, 9]
```

````python
list1 = [[1, 2, 2, 4, 3, 6],
            [5, 1, 3, 4],
            [9, 5, 7, 1],
            [2, 4, 1, 3]]

unique_list = []

for i in list1:
    for j in i:
        if j not in unique_list:
            unique_list.append(j)

print(unique_list)
````

````python
# Another Method
s = set()

for i in list1:
    s.update(i)

print(list(s))
````

### `Q1`: Key with maximum unique values

Given a dictionary with values list, extract key whose value has most unique values.

**Example 1:**

Input:

```bash
test_dict = {"CampusX" : [5, 7, 9, 4, 0], "is" : [6, 7, 4, 3, 3], "Best" : [9, 9, 6, 5, 5]}
```

Output:
```bash
CampusX
```

**Example 2:**

Input:
```bash
test_dict = {"CampusX" : [5, 7, 7, 7, 7], "is" : [6, 7, 7, 7], "Best" : [9, 9, 6, 5, 5]}
```

Output:
```bash
Best
```

````python
test_dict = {"CampusX" : [5, 7, 9, 4, 0], "is" : [6, 7, 4, 3, 3], "Best" : [9, 9, 6, 5, 5]}
lens_lst = []

for i in test_dict:
    lens_lst.append(len(set(test_dict[i])))

for i in test_dict:
    if len(set(test_dict[i])) == max(lens_lst):
        print(i)
````

````python
# Another Method
max_val = 0
max_key = ''

for i in test_dict:
    if max_val < len(set(test_dict[i])):
        max_val = len(set(test_dict[i]))
        max_key = i

print(max_key)
````

### `Q2`: Replace words from Dictionary. Given String, replace it’s words from lookup dictionary.

**Example 1:**

Input:

```bash
test_str = 'CampusX best for DS students.'
repl_dict = {"best" : "is the best channel", "DS" : "Data-Science"}
```

Output:

```bash
CampusX is the best channel for Data-Science students.
```

**Example 2:**

Input:
```bash
test_str = 'CampusX best for DS students.'
repl_dict = {"good" : "is the best channel", "ds" : "Data-Science"}
```

Output:
```bash
CampusX best for DS students.
```

````python
test_str = 'CampusX best for DS students.'
repl_dict = {"best" : "is the best channel", "DS" : "Data-Science"}
output_str = ''

for i in test_str.split():
    if i in repl_dict:
        output_str += repl_dict[i] + " "
    
    else:
        output_str += i + " "

print(output_str)
````

````python
# Another Method
result = []

for i in test_str.split():
    if i in repl_dict:
        result.append(repl_dict[i])
    
    else:
        result.append(i)

print(" ".join(result))
````

### `Q3`: Convert List to List of dictionaries. Given list values and keys list, convert these values to key value pairs in form of list of dictionaries.

**Example 1:**

Input:
```bash
test_list = ["DataScience", 3, "is", 8]
key_list = ["name", "id"]
```

Output:

```bash
[{'name': 'DataScience', 'id': 3}, {'name': 'is', 'id': 8}]
```

**Example 2:**

Input:
```bash
test_list = ["CampusX", 10]
key_list = ["name", "id"]
```

Output:

```bash
[{'name': 'CampusX', 'id': 10}]
```

````python
test_list = ["DataScience", 3, "is", 8]
key_list = ["name", "id"]

test_len = len(test_list)
res = []

for i in range(0, test_len, 2):
    res.append({key_list[0]: test_list[i], key_list[1]: test_list[i+1]})

print(res)
````

### `Q4`: Convert a list of Tuples into Dictionary.

**Example 1:**

Input:
```bash
[("akash", 10), ("gaurav", 12), ("anand", 14), ("suraj", 20), ("akhil", 25), ("ashish", 30)]
```

Output:
```bash
{'akash': [10], 'gaurav': [12], 'anand': [14], 'suraj': [20], 'akhil': [25], 'ashish': [30]}
```

**Example 2:**

Input:
```bash
[('A', 1), ('B', 2), ('C', 3)]

Output:

{'A': [1], 'B': [2], 'C': [3]}
```

````python
list1 = [("akash", 10), ("gaurav", 12), ("anand", 14), ("suraj", 20), ("akhil", 25), ("ashish", 30)]
output_dict = dict()

for i, j in list1:
    output_dict[i] = [j]

print(output_dict) 
````
